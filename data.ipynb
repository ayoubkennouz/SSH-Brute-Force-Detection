```python
# ÉTAPE 3 : EXTRACTION DES FEATURES PAR IP
```

```python
# === ÉTAPE 3.1 : CALCUL DES FEATURES COMPORTEMENTALES ===
from collections import Counter
import math

def calculate_entropy(values):
    """Calcule l'entropie de Shannon d'une liste de valeurs"""
    if len(values) == 0:
        return 0
    counter = Counter(values)
    probabilities = [count / len(values) for count in counter.values()]
    return -sum(p * math.log2(p) for p in probabilities)

# Liste pour stocker les features de chaque IP
features_list = []

# Analyser chaque IP source
for ip, group in df_failures.groupby('src_ip'):
    # Trier par timestamp
    group = group.sort_values('datetime')
    timestamps = group['datetime'].tolist()
    
    # === CALCUL DES FEATURES ===
    
    # Features de volume
    failed_attempts = len(group)
    num_target_servers = group['dst_ip'].nunique()
    num_source_ports = group['src_port'].nunique()
    ssh_client_diversity = group['ssh_client'].nunique()
    
    # Features temporelles
    if len(timestamps) > 1:
        time_diffs = [(timestamps[i+1] - timestamps[i]).total_seconds() 
                     for i in range(len(timestamps)-1)]
        avg_time_between = np.mean(time_diffs)
        time_variance = np.var(time_diffs) if len(time_diffs) > 1 else 0
        total_duration = (timestamps[-1] - timestamps[0]).total_seconds()
    else:
        avg_time_between = 0
        time_variance = 0
        total_duration = 0
    
    # Taux d'activité
    attempts_per_minute = (failed_attempts / (total_duration / 60)) if total_duration > 0 else 0
    
    # Features d'entropie et diversité
    port_entropy = calculate_entropy(group['src_port'].tolist())
    target_entropy = calculate_entropy(group['dst_ip'].tolist())
    
    # Heure de pic d'activité
    hours = [ts.hour for ts in timestamps]
    peak_hour = max(set(hours), key=hours.count) if hours else 0
    
    # === STOCKER LES FEATURES ===
    features = {
        'ip_address': ip,
        'failed_attempts': failed_attempts,
        'num_target_servers': num_target_servers,
        'num_source_ports': num_source_ports,
        'ssh_client_diversity': ssh_client_diversity,
        'avg_time_between_attempts': avg_time_between,
        'time_variance': time_variance,
        'total_duration_minutes': total_duration / 60,
        'attempts_per_minute': attempts_per_minute,
        'port_entropy': port_entropy,
        'target_entropy': target_entropy,
        'peak_hour': peak_hour,
        'first_seen': timestamps[0] if timestamps else None,
        'last_seen': timestamps[-1] if timestamps else None
    }
    
    features_list.append(features)

# Créer le DataFrame des features
df_features = pd.DataFrame(features_list)
print(f"✓ Features calculées pour {len(df_features)} IPs uniques")
```

```python
# === ÉTAPE 3.2 : APERÇU DES FEATURES CALCULÉES ===
print("\n=== APERÇU DES FEATURES ===")
print(f"Nombre d'IPs analysées: {len(df_features)}")
print(f"Colonnes des features: {list(df_features.columns)}")
print(f"\n5 premières IPs analysées:")
print(df_features.head())

print(f"\n=== STATISTIQUES DES FEATURES ===")
print(df_features[['failed_attempts', 'num_target_servers', 'attempts_per_minute']].describe())
```

```python
# === ÉTAPE 4.1 : ÉTIQUETAGE AUTOMATIQUE ===

labels = []

for _, row in df_features.iterrows():
    is_attack = 0
    score = 0
    
    # Système de scoring basé sur le comportement
    # Règle 1: Volume d'échecs
    if row['failed_attempts'] >= 5:
        score += 2
    elif row['failed_attempts'] >= 10:
        score += 3
    elif row['failed_attempts'] >= 50:
        score += 5
    
    # Règle 2: Scanning multiple serveurs
    if row['num_target_servers'] >= 3:
        score += 2
    elif row['num_target_servers'] >= 5:
        score += 3
    
    # Règle 3: Rythme d'attaque
    if row['attempts_per_minute'] >= 2:
        score += 2
    elif row['attempts_per_minute'] >= 5:
        score += 3
    
    # Règle 4: Diversité des ports
    if row['num_source_ports'] >= 3:
        score += 1
    elif row['num_source_ports'] >= 10:
        score += 2
    
    # Règle 5: Entropie élevée (comportement aléatoire)
    if row['port_entropy'] > 2.0:
        score += 1
    
    # Seuil de décision
    if score >= 4:  # Seuil ajustable
        is_attack = 1
    
    labels.append(is_attack)

df_features['is_attack'] = labels
```

```python
# === ÉTAPE 4.2 : ANALYSE DES ÉTIQUETTES ===
attack_count = df_features['is_attack'].sum()
benign_count = len(df_features) - attack_count

print(f"\n=== RÉPARTITION DES ÉTIQUETTES ===")
print(f"IPs étiquetées comme ATTAQUES: {attack_count}/{len(df_features)} ({attack_count/len(df_features)*100:.1f}%)")
print(f"IPs étiquetées comme BÉNIGNES: {benign_count}/{len(df_features)} ({benign_count/len(df_features)*100:.1f}%)")

# Aperçu des IPs étiquetées comme attaques
print(f"\n=== IPS IDENTIFIÉES COMME ATTAQUES ===")
attacking_ips = df_features[df_features['is_attack'] == 1]
print(attacking_ips[['ip_address', 'failed_attempts', 'num_target_servers', 'attempts_per_minute']])
```

```python
# === ÉTAPE 5.1 : CRÉATION DES LÉGITIMES SYNTHÉTIQUES ===

synthetic_data = []
existing_ips = set(df_features['ip_address'].tolist())
base_ip = "192.168.1.{}"

# Créer autant de légitimes que d'attaques pour équilibrer
n_samples = max(attack_count * 2, 20)  # Au moins 20 échantillons

for i in range(n_samples):
    # Générer une IP unique
    while True:
        ip_suffix = np.random.randint(100, 250)
        new_ip = base_ip.format(ip_suffix)
        if new_ip not in existing_ips:
            break
    
    # Comportement légitime typique
    failed_attempts = np.random.choice([0, 1, 2, 3], p=[0.4, 0.3, 0.2, 0.1])
    num_target_servers = 1  # Un seul serveur cible
    num_source_ports = 1   # Un seul port source
    
    # Temps entre tentatives plus long pour les légitimes
    avg_time_between = np.random.uniform(60, 600)  # 1-10 minutes
    
    features = {
        'ip_address': new_ip,
        'failed_attempts': failed_attempts,
        'num_target_servers': num_target_servers,
        'num_source_ports': num_source_ports,
        'ssh_client_diversity': 1,
        'avg_time_between_attempts': avg_time_between,
        'time_variance': np.random.uniform(30, 200),
        'total_duration_minutes': np.random.uniform(2, 15),
        'attempts_per_minute': failed_attempts / np.random.uniform(5, 15),
        'port_entropy': 0,
        'target_entropy': 0,
        'peak_hour': np.random.randint(8, 18),  # Heures de travail
        'first_seen': datetime.now() - timedelta(days=np.random.randint(1, 30)),
        'last_seen': datetime.now(),
        'is_attack': 0
    }
    
    synthetic_data.append(features)

df_synthetic = pd.DataFrame(synthetic_data)
print(f"✓ {len(df_synthetic)} IPs légitimes synthétiques créées")
```

```python
# === ÉTAPE 6.1 : COMBINAISON ET ÉQUILIBRAGE ===

# Combiner les vraies données et les synthétiques
df_final = pd.concat([df_features, df_synthetic], ignore_index=True)

# Mélanger aléatoirement
df_final = df_final.sample(frac=1, random_state=42).reset_index(drop=True)

print(f"\n=== DATASET FINAL ===")
print(f"Taille totale: {len(df_final)} IPs")
final_attack_count = df_final['is_attack'].sum()
final_benign_count = len(df_final) - final_attack_count
print(f"Attaques: {final_attack_count} ({final_attack_count/len(df_final)*100:.1f}%)")
print(f"Légitimes: {final_benign_count} ({final_benign_count/len(df_final)*100:.1f}%)")
```

```python
# === ÉTAPE 6.2 : PRÉPARATION POUR L'ENTRAÎNEMENT ===

# Sélectionner seulement les features numériques pour l'entraînement
features_for_training = [
    'failed_attempts', 'num_target_servers', 'num_source_ports',
    'ssh_client_diversity', 'avg_time_between_attempts', 'time_variance',
    'total_duration_minutes', 'attempts_per_minute', 'port_entropy',
    'target_entropy', 'peak_hour', 'is_attack'
]

df_ready = df_final[features_for_training].copy()

# Remplacer les NaN
df_ready = df_ready.fillna(0)

print(f"\n=== FEATURES POUR ENTRAÎNEMENT ===")
print(f"Colonnes: {list(df_ready.columns)}")
print(f"Shape: {df_ready.shape}")
print(f"\nAperçu:")
print(df_ready.head())
```

```python
# === ÉTAPE 6.3 : SAUVEGARDE FINALE ===

# Sauvegarder le dataset complet
df_final.to_csv('ssh_bruteforce_dataset_complet.csv', index=False)

# Sauvegarder la version pour entraînement (sans métadonnées)
df_ready.to_csv('ssh_bruteforce_training_ready.csv', index=False)

print(f"\n SAUVEGARDE TERMINÉE")
print(f" Dataset complet: ssh_bruteforce_dataset_complet.csv")
print(f" Pour entraînement: ssh_bruteforce_training_ready.csv")
print(f" Taille: {len(df_ready)} échantillons")
print(f"⚖ Équilibre: {final_attack_count} attaques vs {final_benign_count} légitimes")
```
